# Remoção de comentários do projeto
Um notebook que remove todos os comentários de um projeto.

```python
import pathlib
import re
import sys
```

```python
root = pathlib.Path('C:/Users/pepic/Downloads/bootcamp-higao/chirper')
extensions = ['*.php', '*.blade.php', '*.css', '*.js']
files = []
for ext in extensions:
    for path in root.rglob(ext):
        if 'vendor' in path.parts:
            continue
        files.append(path)

print(f'Found {len(files)} files to process.')
```

```python
patterns = [
    re.compile(r'(?s){{--.*?--}}'),
    re.compile(r'(?s)<!--.*?-->'),
    re.compile(r'(?s)/\*.*?\*/'),
    re.compile(r'(?m)^[ \t]*(//|#).*\n?'),
]
modified = 0
for path in files:
    text = path.read_text(encoding='utf-8')
    cleaned = text
    for pattern in patterns:
        cleaned = pattern.sub('', cleaned)
    if cleaned != text:
        path.write_text(cleaned, encoding='utf-8')
        modified += 1

print(f'Modified {modified} files.')
```

```python
comment_patterns = [
    re.compile(r'(?s){{--.*?--}}'),
    re.compile(r'(?s)<!--.*?-->'),
    re.compile(r'(?s)/\*.*?\*/'),
    re.compile(r'(?m)^[ \t]*(//|#).*\n?'),
]
remaining = []
for path in files:
    text = path.read_text(encoding='utf-8')
    if any(pattern.search(text) for pattern in comment_patterns):
        remaining.append(str(path.relative_to(root)))

if remaining:
    print('Comments still found in:')
    for path in remaining:
        print(path)
    sys.exit(1)
else:
    print('No comments found in processed files.')
```
